```python
import ROOT
from myroot import MyTemplatePlotter
```

```python
infile = ROOT.TFile.Open("templates.root")
iron = infile.Get("iron")
copper = infile.Get("copper")
zinc = infile.Get("zinc")
xrf_data1 = infile.Get("xrf_data1")
xrf_data2 = infile.Get("xrf_data2")
```

```python
c1 = ROOT.TCanvas()
```

```python
iron.Draw()
```

```python
c1.Draw()
```

```python
xrf1 = iron.Clone();    xrf1.SetName("xrf1")
xrf_data1.Draw("energy >> xrf1");  c1.Update()
xrf1.Draw("e")  ; c1.Update()
```

```python
xrf2 = iron.Clone();    xrf2.SetName("xrf2")
xrf_data2.Draw("energy >> xrf2");  c1.Update()
xrf2.Draw("e")  ; c1.Update()
```

```python
mtp = MyTemplatePlotter()
```

```python
mtp.AddData(xrf2)
mtp.AddTemplate(iron,'Fe',0.3333*xrf2.Integral())
mtp.AddTemplate(copper,'Cu',0.3333*xrf2.Integral())
mtp.AddTemplate(zinc,'Zn',0.3333*xrf2.Integral())
print(mtp)
```

```python
c2 = ROOT.TCanvas()
mtp.Draw(c2)
```

```python
iron.Scale(1./iron.Integral())
copper.Scale(1./copper.Integral())
zinc.Scale(1./zinc.Integral())
```

```python
def fitall(x,p) :
    e = x[0]
    thebin = iron.FindBin(e)
    return p[0]*(p[1]*iron.GetBinContent(thebin)+p[2]*copper.GetBinContent(thebin)+p[3]*zinc.GetBinContent(thebin))
```

```python
myfit = ROOT.TF1("myfit",fitall,5.,10.,ndim=1,npar=4)
```

```python
myfit.SetParameters(xrf1.Integral(),0.3333,0.3333,0.3333)
myfit.FixParameter(0,xrf1.Integral())
```

```python
xrf1.Fit("myfit","S0")
```

```python
mtpfit = MyTemplatePlotter()
```

```python
mtpfit.AddData(xrf1)
mtpfit.AddTemplate(iron,'Fe',myfit.GetParameter(1)*xrf1.Integral())
mtpfit.AddTemplate(copper,'Cu',myfit.GetParameter(2)*xrf1.Integral())
mtpfit.AddTemplate(zinc,'Zn',myfit.GetParameter(3)*xrf1.Integral())
```

```python
c2 = ROOT.TCanvas()
mtpfit.Draw(c2)
```

```python
iron.Scale(1./iron.Integral())
copper.Scale(1./copper.Integral())
zinc.Scale(1./zinc.Integral())
```
