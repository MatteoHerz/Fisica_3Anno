# Introduzione alla programmazione a oggetti
* È importante creare software in modo rapido, corretto e leggibile
* Gli **oggetti**, o più precisamente, le **classi** da cui provengono gli oggetti, sono essenzialmente _componenti software riutilizzabili_
* Quasi ogni oggetto materiale può essere ragionevolmente rappresentato come un oggetto software in termini di _attributi_ (ad esempio, nome, colore e dimensione) e _azioni_ che può compiere oppure possono essere compiute su di esso (ad esempio, calcolo, movimento e comunicazione)

### Esempio: L'automobile come oggetto
* Prima di poter guidare un'auto, qualcuno deve _progettarla_
* Un'auto di solito inizia con disegni tecnici, simili ai _progetti_ che descrivono la pianta di una casa
* Questi disegni includono il progetto per un pedale dell'acceleratore. Il pedale dell'acceleratore _nasconde_ al conducente i complessi meccanismi che fanno andare l'auto più veloce
* Allo stesso modo, il pedale del freno "nasconde" i meccanismi che rallentano l'auto, il volante "nasconde" i meccanismi che fanno girare l'auto ecc.
* Tutto ciò consente alle persone con poca o nessuna conoscenza di come funzionano i motori, i freni e i meccanismi dello sterzo di guidare un'auto facilmente
* Non è possibile guidare i disegni tecnici di un'auto
* Per prima cosa, bisogna _costruire_ un'auto con i disegni tecnici che la descrivono
* Un'auto completata ha un pedale dell'acceleratore _reale_ per farla andare più veloce quando il conducente lo _preme_

### Metodi e classi
* L'esecuzione di un'attività in un programma richiede un **metodo**, che contiene le istruzioni del programma che eseguono la sua attività
* Il metodo nasconde queste istruzioni al suo utente, proprio come il pedale dell'acceleratore di un'auto nasconde al conducente i meccanismi che fanno andare più veloce l'auto
* Un'unità di programma chiamata **classe** contiene l'insieme di metodi che eseguono le attività della classe
* Una classe che rappresenta un conto bancario potrebbe contenere un metodo per _depositare_ denaro su un conto, un altro per _prelevare_ denaro da un conto e un terzo per _chiedere_ qual è il saldo del conto
* Una classe è simile nel concetto ai disegni tecnici di un'auto

### Istanziazione
* Si deve _costruire un oggetto_ di una classe prima che un programma possa eseguire le attività definite dai metodi della classe
* Questo è chiamato _istanziazione_
* Un oggetto è quindi definito **istanza** della sua classe
```python
   robertos_account = account(initial_value=500)                # classe predefinita, oppure
 
   import bank_software as bs
   robertos_account = bs.account(500)                           # classe nel MODULO (altro file python) bank_software.py 
```

### Riutilizzo
* I disegni tecnici di un'auto possono essere _riutilizzati_ molte volte per costruire molte auto
* Allo stesso modo, puoi _riutilizzare_ una classe molte volte per costruire molti oggetti
* Il riutilizzo di classi esistenti quando si creano nuove classi e programmi fa risparmiare tempo e fatica e aiuta anche a costruire sistemi più affidabili ed efficaci perché le classi e i componenti esistenti sono spesso stati sottoposti a estesi _test_, _debug_ e _ottimizzazione delle prestazioni_.
* In Python, in genere si usa un approccio _a blocchi_ (usare pezzi esistenti di alta qualità ove possibile)

### Messaggi e chiamate di metodo
* Quando si guida un'auto, premendo il pedale dell'acceleratore viene inviato un _messaggio_ all'auto per eseguire un'attività, ovvero per andare più veloce (invio di messaggi a un oggetto)
* Ogni messaggio corrisponde una **chiamata di metodo** che indica a un metodo dell'oggetto di eseguire la sua attività
  * Ad esempio, si chiama il metodo `deposit` di un oggetto conto bancario per aumentare il saldo del conto
```python
   new_value = robertos_account.deposit(800)
   print (new_value)
   > 1300
```

### Attributi e variabili di istanza
* Un'auto ha anche degli _attributi_, come il colore, il numero di porte, la quantità di benzina nel serbatoio, la velocità attuale e il record dei chilometri totali percorsi (ad esempio, la lettura del contachilometri)
* Gli attributi sono rappresentati come parte del suo design nei suoi disegni tecnici
* Quando si ha a che fare con un'auto vera e propria, questi attributi sono parte integrante dell'auto
* Ogni auto mantiene i suoi _propri_ attributi, ad esempio, ogni auto sa quanta benzina c'è nel suo serbatoio, ma _non_ quanta ce n'è nei serbatoi delle _altre_ auto

* Un oggetto ha attributi che porta con sé quando viene utilizzato in un programma, specificati come parte della classe dell'oggetto
    * Un oggetto conto bancario ha un _attributo saldo_ che rappresenta la quantità di denaro nel conto
    * Ogni oggetto conto bancario conosce il saldo nel conto che rappresenta, ma _non_ i saldi degli _altri_ conti nella banca
* Gli attributi sono specificati dalle **variabili di istanza** della classe
* Gli attributi e i metodi di una classe sono intimamente correlati, quindi le classi racchiudono insieme i loro attributi e metodi
```python
   print (robertos_account.balance)                            ## SENZA parentesi tonde aperte/chiuse
   > 1300
```   

### L'esempio del nostro programma
```python 
figure = plt.figure(figsize=(6,8))  ### ISTANZIAZIONE di un oggetto di classe "figure" dal file "plt" che corrisponde a matplotlib/pyplot.py 

ax_top = figure.add_subplot(211)    ### CHIAMATA DI METODO della classe "figure" che come risultato produce l'ISTANZIAZIONE di un oggetto di classe "axis" 

ndecay_vs_time = ax_top.bar(times,n_decays)    ### CHIAMATA DI METODO della classe "axis" che come risultato produce l'ISTANZIAZIONE di un oggetto di classe "bar_container" che ha come ATTRIBUTI le liste 'times' e 'n_decays'

ax_top.grid(True)                              ### 4 CHIAMATE DI METODO della classe "axis" che modificano altri ATTRIBUTI dell'oggetto "ax_top" di tipo "axis", per es. l'etichetta dell'asse x da '' (valore di default della VARIABILE DI ISTANZA) diventa 'Time interval [min]'  

ax_top.set_xlabel('Time interval [min]')
ax_top.set_ylabel('Decayed nuclei in time interval')
ax_top.set_title('Radioactive decays')
```


### Ereditarietà
* Una nuova classe di oggetti può essere creata comodamente tramite **ereditarietà**
* La nuova classe (chiamata **sottoclasse**) inizia con le caratteristiche di una classe esistente (chiamata **superclasse**), eventualmente personalizzandole e aggiungendo caratteristiche uniche proprie

### Analisi e progettazione orientate agli oggetti (OOAD)
* Linguaggi come Python sono orientati agli oggetti. La programmazione in un tale linguaggio, chiamata **programmazione orientata agli oggetti (OOP)**, consente di implementare un design orientato agli oggetti nel programmare.
* I progetti grandi e complessi dovrebbero seguire un processo di **analisi** dettagliato per determinare i **requisiti** del progetto (_cosa_ dovrebbe fare il sistema), quindi sviluppare un **design** che li soddisfi (_come_ il sistema dovrebbe farlo), prima di scrivere qualsiasi codice
* Se questo processo comporta l'analisi e la progettazione del sistema da un punto di vista orientato agli oggetti, si chiama processo di **analisi e progettazione orientate agli oggetti (OOAD)**
