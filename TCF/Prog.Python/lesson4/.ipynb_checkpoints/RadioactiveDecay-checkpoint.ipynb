```python
"""Dynamically graphing frequencies of radioactive decays."""
#import matplotlib
#matplotlib.use('TkAgg')
from matplotlib import animation
import matplotlib.pyplot as plt
import random 
import math

# number of animation frames
number_of_frames = 1000 

# radioactive-isotope half-life
half_life = 155  # in minutes

# number of radioactive-isotope nuclei 
n0_nuclei = 200_000

def init(): 
    global n_nuclei
    n_nuclei = n0_nuclei

def update(frame_number, n_decays, frequencies):
    """Configures bar plots contents for each animation frame."""
    global n_nuclei
    # simulate number of nuclei decayed between t and t + Deltat
    n_decays_in_time = [random.uniform(0.,1.) < 1-math.exp(-1./half_life) for i in range(n_nuclei)].count(True)
    n_decays[frame_number] = n_decays_in_time
    n_nuclei -= n_decays_in_time
    print ("N_nuclei left = ", n_nuclei)
    
    # generate frequencies
    if (n_decays_in_time < len(frequencies)) :
        frequencies[n_decays_in_time] += 1
    
    for bars, h in zip(ndecay_vs_time,n_decays):
        bars.set_height(h)
    ax_top.set_ylim(0., max(n_decays)*1.1)    
    for bars2, h2 in zip(frequency_of_ndecay,frequencies):
        bars2.set_height(h2)
    ax_bottom.set_ylim(0., max(frequencies)*1.1)   

# Ndecays vs. time plot
times    = list(range(number_of_frames))    # 1000 one-minute consecutive intervals
n_decays = [0.] * number_of_frames          # number of decayed nuclei in interval i

# frequency of Ndecays
max_freq = min(500, 4 * n0_nuclei // half_life)
n_decays_value = list(range(max_freq))              # possible ndecays values, conservatively up to 4*average (average = N_nuclei * Delta t/tau)  
frequencies   = [0] * max_freq                      # frequency of occurrence of one specific ndecays value

figure = plt.figure(figsize=(6,8))
ax_top = figure.add_subplot(211)             # con '.add_subplot()' specifico che nel mio .figure ci saranno più sottografici diversi
ndecay_vs_time = ax_top.bar(times,n_decays)  # '.bar' istogramma a cui passo (x points, y values)
ax_top.grid(True)                            # "ax" indica un piano cartesiano e non un solo asse su Python
ax_top.set_xlabel('Time interval [min]')
ax_top.set_ylabel('Decayed nuclei in time interval')
ax_top.set_title('Radioactive decays')

ax_bottom = figure.add_subplot(212)
frequency_of_ndecay = ax_bottom.bar(n_decays_value,frequencies)
ax_bottom.grid(True)
ax_bottom.set_xlabel('Decayed nuclei in time interval')
ax_bottom.set_ylabel('Frequency')

# configure and start animation that calls function update
rd_animation = animation.FuncAnimation(
    figure, update, init_func=init, repeat=False, frames=number_of_frames,
    fargs=(n_decays, frequencies))

plt.show()  # display window
```
