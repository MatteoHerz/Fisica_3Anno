# Visualizzazioni dinamiche
* La funzione **`FuncAnimation`** del modulo **`animation`** di Matplotlib aggiorna una visualizzazione _dinamicamente_. 

### Frame di animazione
* `FuncAnimation` implementa una **animazione frame per frame**. 
* Ogni **frame di animazione** specifica cosa cambiare in un singolo aggiornamento dei grafici. 
* Molti aggiornamenti rapidi nel tempo creano un'animazione. 
* Questo esempio (decadimento radioattivo) non ha definito un intervallo fisso di frame al secondo, che quindi è fissato dal tempo di esecuzione di un singolo aggiornamento.

* `FuncAnimation` chiama la funzione `update` **una volta per frame**. 
* Parametri:
    * `frame_number`—Il valore successivo dell'argomento `frames` di `FuncAnimation` che varia da 0 fino al totale dei frame definiti. Questo argomento è obbligatorio e ha sempre questo significato. I successivi parametri sono specifici dell'animazione e opzionali.

### 4) Configurazione di dati e grafici
* Vengono definite liste di numeri interi (per gli assi delle ascisse dei due grafici, nel primo caso intervalli di 1 minuto, nel secondo possibili valori di `n_decays` in uno di questi intervalli) e liste vuote di lunghezza corrispondente—notare sintassi utilizzata.
* Viene definita una figura e aggiunti i `subplot` a mano con l'argomento (_xyz_), dove _x_ = numero di righe totale, _y_ = numero di colonne totale, _z_ = posizione di questo subplot. In ogni subplot viene istanziato un grafico a barre, detto `bar` in Matplotlib, dove i due argomenti sono le ascisse e le altezze delle barre (queste ultime tutte nulle inizialmente). 

### Funzione `update`: cosa avviene in ogni frame dell'animazione 
* **Simulazione** dei decadimenti dei nuclei **restanti** _n_nuclei_ al frame considerato.
* Ognuno di essi ha probabilità $1-e^{-\Delta t/\tau}$ di decadere in $\Delta t$. Si contano quanti sono nell'intervallo considerato e si aumenta di 1 la frequenza corrispondente a quel numero.
* Si ricalcolano i nuclei radioattivi restanti sottraendo il numero di quelli decaduti in $\Delta t$.

```python
    global n_nuclei
    # simulate number of nuclei decayed between t and t + Deltat
    n_decays_in_time = [random.uniform(0.,1.) < 1-math.exp(-1./half_life) for i in range(n_nuclei)].count(True)
    n_decays[frame_number] = n_decays_in_time
    n_nuclei -= n_decays_in_time
    print ("N_nuclei left = ", n_nuclei)
    
    # generate frequencies
    if (n_decays_in_time < len(frequencies)) :
        frequencies[n_decays_in_time] += 1
```
* Si aggiornano le altezze dei grafici corrispondenti e allo stesso momento i limiti dell'asse y perché il grafico vi sia sempre contenuto.
```python
    for bars, h in zip(ndecay_vs_time,n_decays):
        bars.set_height(h)
    ax_top.set_ylim(0., max(n_decays)*1.1)    
    for bars2, h2 in zip(frequency_of_ndecay,frequencies):
        bars2.set_height(h2)
    ax_bottom.set_ylim(0., max(frequencies)*1.1)   
```
* **NOTA BENE:**
  * La notazione `zip()` per il ciclo su due iterabili allo stesso momento.
  * quando `bar` è usato come **iterabile** restituisce una **lista** delle barre del grafico.
  * `set_ylim` risetta l'intervallo dell'asse y il cui massimo diventa sempre il valore più alto tra le barre aumentato del 10%.

### 5) Chiamata a `FuncAnimation` 
* `FuncAnimation` restituisce un oggetto che _deve_ essere definito anche se non viene usato (altrimenti finisce nella raccolta dei rifiuti). 

```python
# configure and start animation that calls function update
rd_animation = animation.FuncAnimation(
    figure, update, init_func=init, repeat=False, frames=number_of_frames,
    fargs=(n_decays, frequencies))

plt.show()  # display window

```

`FuncAnimation` ha due parametri obbligatori:
* `figure`—l'oggetto `Figure` dove mostrare l'animazione e
* `update`—la funzione da chiamare ad ogni frame.

* Argomenti opzionali:
    * **`repeat`**—Se `True` (default), l'animazione riprende dall'inizio una volta completata.
    * **`frames`**—Il numero totale di frames, che controlla quante volte `FunctAnimation` chiama `update`. 
    * **`init_func`**—Il nome della funzione inizializzazione.
    * **`fargs`** (cioè  “function arguments”)—Una tupla di altri argomenti (a parte quello obbligatorio, ovvero il numero di frame, che **non va** elencato) da passare alla funzione `update` di `FuncAnimation`. 
* [Altri argomenti opzionali](https://matplotlib.org/api/_as_gen/matplotlib.animation.FuncAnimation.html)
