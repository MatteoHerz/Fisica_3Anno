# Variabili locali e globali
* Ogni identificatore ha un intervallo di validità (`scope`) che determina **dove** può essere usato nel programma.
* [Dettagli completi sugli scope e spazi dei nomi](https://docs.python.org/3/tutorial/classes.html#python-scopes-and-namespaces).

* Una variabile definita per esempio all'interno di una suite o di una funzione ha un intervallo di **validità locale**.
* Un identificatore definito al di fuori di ogni funzione o classe ha **validità globale**—può essere una variabile, una funzione o una classe.

### Come accedere a una variabile globale in una funzione

```python
x = 7
```

```python
def access_global():
    print('x printed from access_global:', x)
```

```python
access_global()
```

* Di default, non si può _modificare_ una variabile globale in una funzione.
* Provandoci, Python crea invece una **nuova variable locale** quando viene assegnato un valore a quella variabile all'interno della funzione.
* Nel blocco `try_to_modify_global`, la `x` locale **adombra** la `x` globale, rendendola inaccessibile all'interno della funzione. 

```python
def try_to_modify_global():
    x = 3.5
    print('x printed from try_to_modify_global:', x)
```

```python
try_to_modify_global()
```

```python
x
```

* Per modificare una variabile globale dentro una funzione è necessario usare esplicitamente un'istruzione **`global`**:

```python
def modify_global():
    global x;
    x = 'hello'
    print('x printed from modify_global:', x)
```

```python
modify_global()
```

```python
x
```

### Adombramento 
* ATTENZIONE: Variabili definite con lo stesso nome di funzioni possono adombrarle e renderle inutilizzabili! Ecco cosa succede definendo una variabile di nome `sum`. 

```python
sum = 10 + 5
```

```python
sum
```

```python
sum([10, 5])
```
