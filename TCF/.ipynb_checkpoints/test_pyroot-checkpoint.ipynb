```python
# Importiamo pyROOT
import ROOT

# Creiamo un istogramma
h = ROOT.TH1F("h1", "Istogramma di prova", 100, -5, 5)

# Riempimento con numeri casuali Gaussiani
for i in range(1000):
    h.Fill(ROOT.gRandom.Gaus(0, 1))

# Creiamo un canvas per visualizzare
c = ROOT.TCanvas("c1", "Canvas di prova", 800, 600)
h.Draw()
c.Draw()
```

```python
# Controlliamo la versione di ROOT
print("ROOT version:", ROOT.gROOT.GetVersion())
```
