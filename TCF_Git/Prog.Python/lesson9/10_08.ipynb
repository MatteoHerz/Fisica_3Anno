# Creazione di una gerarchia di ereditarietà. Introduzione al polimorfismo
* Esempio: la gerarchia contenente le tipologie di dipendenti nell'app di gestione stipendi di un'azienda
* Tutti i dipendenti dell'azienda hanno molto in comune
* I _dipendenti a provvigione_ (che saranno rappresentati come oggetti di una **classe base**) ricevono una percentuale sulle loro vendite
* I _dipendenti a provvigione_ stipendiati (che saranno rappresentati come oggetti di una **sottoclasse**) ricevono una percentuale sulle loro vendite _più_ uno stipendio base


## Classe base `CommissionEmployee`
La classe `CommissionEmployee` offre le seguenti funzionalità:
* Il metodo `__init__` crea gli attributi `_first_name`, `_last_name` e `_ssn` (codice fiscale) e utilizza i setter delle proprietà `gross_sales` e `commission_rate` per creare i corrispondenti attributi.
* Proprietà di lettura e scrittura `commission_rate` in cui il setter esegue la convalida dei dati.
* Metodo `earnings`, che calcola e restituisce i guadagni di un `CommissionEmployee`.
* Metodo `__str__`, che restituisce una rappresentazione in formato stringa di un `CommissionEmployee`.

````python
# commmissionemployee.py
"""CommissionEmployee base class."""

class CommissionEmployee:
    """An employee who gets paid commission based on gross sales."""

    def __init__(self, first_name, last_name, ssn, 
                 gross_sales, commission_rate):
        """Initialize CommissionEmployee's attributes."""
        self._first_name = first_name
        self._last_name = last_name
        self._ssn = ssn
        self._gross_sales = gross_sales  # validate via property
        self.commission_rate = commission_rate  # validate via property
        
    @property
    def commission_rate(self):
        return self._commission_rate

    @commission_rate.setter
    def commission_rate(self, rate):
        """Set commission rate or raise ValueError if invalid."""
        if not (0. < rate < 1.):
            raise ValueError(
               'Interest rate must be greater than 0 and less than 1')
        
        self._commission_rate = rate

    def earnings(self):
        """Calculate earnings."""   
        return self._gross_sales * self.commission_rate

    def __str__(self):
        """Return string representation for str()."""
        return ('CommissionEmployee: ' + 
            f'{self._first_name} {self._last_name}\n' +
            f'social security number: {self._ssn}\n' +
            f'gross sales: {self._gross_sales:.2f}\n' +
            f'commission rate: {self.commission_rate:.2f}')
````

### Tutte le classi ereditano direttamente o indirettamente dalla classe `object`
* _Ogni_ classe Python eredita da una classe esistente
* Quando non si specifica esplicitamente la classe base per una nuova classe, Python presuppone che la classe erediti direttamente dalla classe `object`
* L'intestazione della classe `CommissionEmployee` avrebbe potuto essere scritta come
```python
class CommissionEmployee(object):
```
* Le parentesi dopo `CommissionEmployee` indicano l'ereditarietà e possono contenere:
   * una singola classe per l'ereditarietà singola
   * un elenco di classi base separate da virgole per l'ereditarietà multipla

* `CommissionEmployee` eredita tutti i metodi della classe `object`
* Due dei numerosi metodi ereditati da `object` sono `__init__` e `__str__`
* Quindi _ogni_ classe ha il metodo che restituisce rappresentazioni stringa dell'oggetto su cui vengono chiamate
* Quando l'implementazione di un metodo della classe base non è appropriata per una classe derivata, tale metodo può essere **sovrascritto** (ovvero ridefinito) nella classe derivata con un'implementazione appropriata
* Il metodo `__str__` sovrascrive l'implementazione predefinita della classe `object`

### Proviamo la classe `CommissionEmployee`  

````python
c = CommissionEmployee('Sue', 'Jones', '333-33-3333', 10000.00, 0.06)
````

````python
print(c)
````

````python
print(f'{c.earnings():,.2f}')
````

````python
c.commission_rate = 1.5
````

````python
c.commission_rate = 0.1
````

````python
print(f'{c.earnings():,.2f}')
````

## La sottoclasse `SalariedCommissionEmployee`
* Con l'ereditarietà singola, la sottoclasse inizia essenzialmente come la classe base.
* Il vero punto di forza dell'ereditarietà deriva dalla possibilità di definire nella sottoclasse aggiunte, sostituzioni o perfezionamenti per le funzionalità ereditate dalla classe base.
* Molte delle funzionalità di `SalariedCommissionEmployee` sono simili, se non identiche, a quelle della classe `CommissionEmployee`.
* Entrambi i tipi di dipendenti hanno attributi relativi a nome, cognome, codice fiscale, fatturato lordo e commissioni, nonché proprietà e metodi per manipolare tali dati.
* L'ereditarietà ci consente di "assorbire" le funzionalità di una classe _senza_ duplicare il codice.

### Dichiarazione della classe `SalariedCommissionEmployee`
* La sottoclasse `SalariedCommissionEmployee` _eredita_ la maggior parte delle sue funzionalità dalla classe `CommissionEmployee`
* Una classe `SalariedCommissionEmployee` _è una_ classe `CommissionEmployee` (perché l'ereditarietà trasferisce le funzionalità della classe `CommissionEmployee`)
* La classe `SalariedCommissionEmployee` presenta inoltre le seguenti caratteristiche:
   * Metodo `__init__`, che inizializza tutti i dati ereditati dalla classe `CommissionEmployee` più un un attributo detto `_base_salary`
   * Una versione personalizzata del metodo  `earnings`
   * Una versione personalizzata del metodo `__str__`

````python
# salariedcommissionemployee.py
"""SalariedCommissionEmployee derived from CommissionEmployee."""

class SalariedCommissionEmployee(CommissionEmployee):
    """An employee who gets paid a salary plus 
    commission based on gross sales."""

    def __init__(self, first_name, last_name, ssn, 
                 gross_sales, commission_rate, base_salary):
        """Initialize SalariedCommissionEmployee's attributes."""
        super().__init__(first_name, last_name, ssn, 
                         gross_sales, commission_rate)
        self._base_salary = base_salary  # validate via property

    def earnings(self):
        """Calculate earnings."""   
        return super().earnings() + self._base_salary

    def __str__(self):
        """Return string representation for str()."""
        return ('Salaried' + super().__str__() +      
            f'\nbase salary: {self._base_salary:.2f}')
````

### Come ereditare dalla classe `CommissionEmployee`

```python
class SalariedCommissionEmployee(CommissionEmployee):
```
* specifica che la classe `SalariedCommissionEmployee` **eredita** da `CommissionEmployee`
* Non vedo gli attributi, le proprietà e i metodi della classe `CommissionEmployee` nella classe `SalariedCommissionEmployee`, ma sono presenti

### Metodo `__init__` e funzione predefinita `super`
* _Ogni sottoclasse `__init__` deve chiamare esplicitamente il metodo `__init__` della sua classe base per inizializzare gli attributi dati ereditati dalla classe base_. Questa chiamata deve essere la prima istruzione nel metodo `__init__` della sottoclasse
* La notazione `super().__init__` utilizza la funzione predefinita **`super`** per individuare e chiamare il metodo `__init__` della classe base che, in relazione alla sottoclasse, è detta **sovraclasse** 

### Sovrascrittura del metodo `earnings`
* Il metodo `earnings` della classe `SalariedCommissionEmployee` sovrascrive il metodo `earnings` della classe `CommissionEmployee` per calcolare gli utili di un `SalariedCommissionEmployee`
* Ottiene la quota degli utili basata sulle _solo commissioni_ chiamando il metodo `earnings` di `CommissionEmployee` con l'espressione `super().earnings()`
* Una cosa simile avviene per il metodo `__str__`.

### Proviamo la classe `SalariedCommissionEmployee` 

````python
s = SalariedCommissionEmployee('Bob', 'Lewis', '444-44-4444',
        5000.00, 0.04, 300.00)
````

````python
print(s)
````

* L'oggetto `SalariedCommissionEmployee` ha _tutte_ le proprietà delle classi `CommissionEmployee` _e_ `SalariedCommissionEmployee`
* Calcola e visualizza i guadagni di `SalariedCommissionEmployee`

````python
print(f'{s.earnings():,.2f}')
````

````python
s.commission_rate = 0.05
````

````python
print(s)
````

````python
print(f'{s.earnings():,.2f}')
````

### Test della relazione "è un/una"
Le funzioni **`issubclass`** e **`isinstance`** vengono utilizzate per testare le relazioni "è un/una".
* ``issubclass` determina se una classe è derivata da un'altra.

````python
issubclass(SalariedCommissionEmployee, CommissionEmployee)
````

* `isinstance` determina se un oggetto ha una relazione "è un/una" con un tipo specifico

````python
isinstance(s, CommissionEmployee)
````

````python
isinstance(s, SalariedCommissionEmployee)
````

Elaborazione polimorfica di `CommissionEmployee` e `SalariedCommissionEmployee`
* Con l'ereditarietà, ogni oggetto di una sottoclasse può essere trattato anche come un oggetto della classe base di quella sottoclasse
* È possibile sfruttare questa relazione per inserire oggetti correlati tramite ereditarietà in un elenco, quindi scorrere l'elenco e trattare ogni elemento come un oggetto della classe base
* Consente l'elaborazione di una varietà di oggetti in modo _generale_

````python
employees = [c, s]
````

````python
for employee in employees:
    print(employee)
    print(f'{employee.earnings():,.2f}\n')
````

* Per ogni dipendente vengono visualizzati correttamente la rappresentazione delle stringhe e i relativi guadagni.
* Questo è chiamato **polimorfismo**, una capacità fondamentale della programmazione orientata agli oggetti (OOP).
