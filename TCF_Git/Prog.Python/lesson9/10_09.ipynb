# I decoratori
* Abbiamo già visto un esempio di **decoratore** per sovrascrivere un setter, cioè un'istruzione di assegnazione
* Possiamo ora programmare dei decoratori personalizzati. 

## Definizione generale
* In Python, i decoratori rappresentano un modo potente e flessibile per modificare o estendere il comportamento di funzioni o metodi, senza modificarne il codice. Un decoratore è essenzialmente una funzione che accetta un'altra funzione come argomento e restituisce una nuova funzione con funzionalità migliorate.
* I decoratori sono spesso utilizzati in scenari come il logging, l'autenticazione e la memorizzazione, consentendo di aggiungere funzionalità aggiuntive a funzioni o metodi esistenti in modo pulito e riutilizzabile.

## Sintassi 

```python
def decorator_name(func):
    def wrapper(*args, **kwargs):
        # Add functionality before the original function call
        result = func(*args, **kwargs)
        # Add functionality after the original function call
        return result
    return wrapper

@decorator_name
def function_to_decorate():
    # Original function code
    return
```

1. decorator_name(func):

* **decorator_name**: Questo è il nome della funzione decoratore.
* **func**: Questo parametro rappresenta la funzione decorata. Quando si utilizza un decoratore, la funzione decorata viene passata a questo parametro.

2. wrapper(*args, **kwargs):

* **wrapper**: Questa è una funzione annidata all'interno del decoratore. Incapsula la funzione originale, aggiungendo funzionalità aggiuntive.
   * *args: Raccoglie tutti gli argomenti posizionali passati alla funzione decorata in una tupla.
   * **kwargs: Raccoglie tutti gli argomenti per parola chiave passati alla funzione decorata in un dizionario.
* La funzione wrapper consente al decoratore di gestire funzioni con qualsiasi numero e tipo di argomenti.

3. @decorator_name:

* Questa sintassi applica il decoratore alla funzione function_to_decorate. Equivale a scrivere:
```python
   function_to_decorate = decorator_name(function_to_decorate).
```
