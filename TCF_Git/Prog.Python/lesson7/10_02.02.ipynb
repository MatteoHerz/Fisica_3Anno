## Definizione di una nuova classe 
* La definizione di una classe inizia con la parola chiave **`class`** seguita dal nome e dai due punti (`:`)
* Questa si chiama **intestazione della classe** (**class header**)
* Ogni istruzione nella suite della classe deve essere indentato

```python
# account.py

class Account:                   # oppure   class Account(object):     (v. sotto)
    """Account class for maintaining a bank account balance."""
    

```

* Ogni classe ha tipicamente una stringa di documentazione che appare subito dopo l'intestazione
* Per vedere la _docstring_ in IPython/Jupyter, si può inserire il nome della classe seguito da un punto interrogativo e poi _Invio_

````python
class Account:
    """Account class for maintaining a bank account balance."""

Account?
````

* In Python `Account` è sia il nome della classe sia quello usato dal costruttore per istanziare un nuovo oggetto chiamando il metodo `__init__`


### Inizializzazione: il metodo `__init__` 
* L'espressione del **costruttore** indica quando, _utilizzando_ la classe in un altro programma si istanzia un oggetto di tipo `Account`:
```python
import account
my_account = account.Account('Roberto',500)
```
* Questo equivale sempre alla chiamata del metodo **`__init__`** che specifica come inizializzare gli attributi di dati dell'oggetto
* Questo metodo deve sempre ritornare `None`, altrimenti si verifica un `TypeError`
* Il metodo `__init__` inizializza un oggetto `Account` con gli attributi `name` e `balance` nel caso il secondo sia un valore valido


```python
    def __init__(self, name, balance):
        """Initialize an Account object."""

        # if balance is less than 0.00, raise an exception
        if balance < 0.:
            raise ValueError('Initial balance must be >= to 0.00.')

        self.name = name
        self.balance = balance


```

### Inizializzazione: il metodo `__init__` (altre informazioni)
* Quando viene chiamato un metodo di un oggetto, Python implicitamente passa una referenza all'oggetto in questione come _primo argomento_ del metodo
* Quindi tutti i metodi di una classe devono avere almeno un argomento, per convenzione questo primo parametro si indica con `self`
* I metodi della classe devono usare l'oggetto `self` per accedere agli attributi e agli altri metodi della classe stessa

* Quando si crea un oggetto, di default non ha attributi, si aggiungono _dinamicamente_ con istruzioni di assegnazione come:
```python
self.attribute_name `=` value
``` 

* Nelle classi di Python si possono usare diversi [metodi speciali](https://docs.python.org/3/reference/datamodel.html#special-method-names), come `__init__`
* Si riconoscono dal doppio underscore (`__`) prima e dopo il nome
* Tutte le classi per default ereditano dalla classe di base **`object`** che definisce i metodi speciali disponibili per _tutti_ gli oggetti Python 

### Un altro esempio di metodo `deposit` 
* Aggiunge un valore positivo `amount` all'attributo `balance` del conto
* Genera un `ValueError` se `amount` è negativo

```python
    def deposit(self, amount):
        """Deposit money to the account."""

        # if amount is less than 0.00, raise an exception
        if amount < 0.:
            raise ValueError('amount must be positive.')

        self.balance += amount

```
