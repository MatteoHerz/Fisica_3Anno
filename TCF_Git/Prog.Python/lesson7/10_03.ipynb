# Incapsulamento 
* Il precedente esempio usava gli attributi `name` e `balance` per _accedere_ ai valori di questi attributi
* In genere è possibile usare gli stessi attributi per _modificare_ tali valori

```python
class Account:
    """Account class for maintaining a bank account balance."""

    def __init__(self, name, balance):
        """Initialize an Account object."""

        # if balance is less than 0.00, raise an exception
        if balance < 0.:
            raise ValueError('Initial balance must be >= to 0.00.')

        self.name = name
        self.balance = balance

    def deposit(self, amount):
        """Deposit money to the account."""

        # if amount is less than 0.00, raise an exception
        if amount < 0.:
            raise ValueError('amount must be positive.')

        self.balance += amount
```

```python
account1 = Account('Roberto', 50.00)
```

```python
account1.balance
```

* Proviamo a modificare il valore di `balance` settando un valore negativo _invalido_ e poi a mostrare `balance`

```python
account1.balance = -1000.00
```

```python
account1.balance
```

### Incapsulamento
* Il **codice cliente** è il codice che usa gli oggetti della classe (metodi e attributi)
* Molti linguaggi di programmazione orientati agli oggetti permettono di **incapsulare** (ovvero _nascondere_) i dati interni alla classe al codice cliente (es. dati _private_ in C++)

### Convenzione del trattino basso iniziale (`_`) 
* **Non esistono** dati privati in Python 
* L'uso corretto va suggerito tramite _convenzioni di denominazione_ 
* Per convenzione, in Python i nomi di attributi che iniziano con un trattino basso (`_`) sono unicamente per _uso interno alla classe_
* Invece i nomi di attributi che **non** iniziano con un trattino basso (`_`) sono _accessibili pubblicamente_ al codice cliente
