### 3) Fit simultaneo (o combinato)

```python
# Ci sono 11 (!!!) parametri di fit 
# la forma/quantità del fondo e la quantità di segnale sono ESCLUSIVI di ogni modo
  ## modo g: 0, 1, 2
  ## modo v: 3, 4, 5
  ## modo a: 6, 7, 8
# la massa media e la risoluzione o dev. standard sono COMUNI a tutti e tre (9, 10) 
import numpy as np

ipar_per_mode = [
    np.array([0, 1, 2, 9, 10], dtype=np.int32),
    np.array([3, 4, 5, 9, 10], dtype=np.int32),
    np.array([6, 7, 8, 9, 10], dtype=np.int32)
]

class GlobalChi2(object):
    def __init__(self, fg, fv, fa):
        self._fg = fg
        self._fv = fv
        self._fa = fa
 
    def __call__(self, par):       
        par_arr = np.frombuffer(par, dtype=np.float64, count=11)
 
        pg = par_arr[ipar_per_mode[0]]
        pv = par_arr[ipar_per_mode[1]]
        pa = par_arr[ipar_per_mode[2]]
 
        return self._fg(pg) + self._fv(pv) + self._fa(pa)
```

```python
expgaus_per_mode = []
wrapped_funcs = []

for i in range(3) : 
  this_func = expgaus.Clone()
  expgaus_per_mode.append(this_func)
  this_wf = ROOT.Math.WrappedMultiTF1(this_func, 1)  # dimension of expgaus (here 1-dim)
  wrapped_funcs.append(this_wf)  
```

```python
opt = ROOT.Fit.DataOptions()
therange = ROOT.Fit.DataRange()
# set the data range
therange.SetRange(95.,155.)
data_binned = [ROOT.Fit.BinData(opt, therange), ROOT.Fit.BinData(opt, therange), ROOT.Fit.BinData(opt, therange)] 
```

```python
chi_squares = []

for i in range(3): 
   ROOT.Fit.FillData(data_binned[i], h_inv_mass[i])
   this_chisq = ROOT.Fit.Chi2Function(data_binned[i], wrapped_funcs[i]) 
   chi_squares.append(this_chisq)
```

```python
globalChi2 = GlobalChi2(chi_squares[0], chi_squares[1], chi_squares[2])
```

```python
fitter = ROOT.Fit.Fitter()
```

```python
par_init = np.array([12000., 0.019, 295., 4000., 0.016, 100., 300., 0.013, 47., 125., 2.])
```

```python
fitter.Config().SetParamsSettings(11, par_init)
```

```python
fitter.Config().ParSettings(0).SetLimits(2000., 3.0e4)
fitter.Config().ParSettings(2).SetLimits(0., 500.)
fitter.Config().ParSettings(3).SetLimits(500., 10000.)
fitter.Config().ParSettings(5).SetLimits(0., 300.)
fitter.Config().ParSettings(6).SetLimits(50., 1000.)
fitter.Config().ParSettings(8).SetLimits(0., 100.)
fitter.Config().ParSettings(9).SetLimits(120., 130.)
fitter.Config().ParSettings(10).SetLimits(0.5, 4.)
for i in {1,4,7} : 
  fitter.Config().ParSettings(i).SetLimits(0.005, 0.03)
```

```python
fitter.Config().MinimizerOptions().SetPrintLevel(1)
fitter.Config().SetMinimizer("Minuit2", "Migrad")
```

```python
globalChi2Functor = ROOT.Math.Functor(globalChi2, 11)
alldata = sum([data_binned[i].Size() for i in range(3)])
fitter.FitFCN(globalChi2Functor, 0, alldata, True)
```

```python
c_cool_fit = ROOT.TCanvas("c_cool_fit", "Simultaneous fit of three histograms", 1200, 400)
c_cool_fit.Divide(3, 1)
```

```python
result = fitter.Result()
```

```python
for i in range(3):
  c_cool_fit.cd(i+1)
  expgaus_per_mode[i].SetFitResult(result, ipar_per_mode[i])
  h_inv_mass[i].GetListOfFunctions().Add(expgaus_per_mode[i])
  h_inv_mass[i].Draw("E")
c_cool_fit.Draw()
```

### 4) Una misura importante!

* Sarete certamente invitati a un convegno a presentarla! Non possiamo mostrare questi grafici orrendi.

```python
c_conference = ROOT.TCanvas("c_conference", "Simultaneous fit of three histograms", 1200, 400)
c_conference.Divide(3,1)
```

```python
ROOT.gStyle.SetOptTitle(0)
ROOT.gStyle.SetOptStat(0)
```

```python
onlyexp = [ROOT.TF1()] * 3

for i in range(3):
   c_conference.cd(i+1)
   ROOT.gPad.SetTicky(1)                  # stanghettine dell'asse anche sopra
   ROOT.gPad.SetTickx(1)                  # stanghettine dell'asse anche a destra 
   ROOT.gPad.SetTopMargin(0.02)           # margini intorno al plot (sx, dx, alto, basso) in  unità NDC (Natural Device Coordinates) - da 0 a 1 nel TPad
   ROOT.gPad.SetRightMargin(0.07)
   ROOT.gPad.SetLeftMargin(0.2)
   ROOT.gPad.SetBottomMargin(0.15)
   h_inv_mass[i].SetLineColor(ROOT.kBlack)           # Colori di linee e puntini
   h_inv_mass[i].SetMarkerColor(ROOT.kBlack)
   h_inv_mass[i].SetMarkerStyle(20)                  # Forma e dimensione del puntino
   h_inv_mass[i].SetMarkerSize(0.7)
   h_inv_mass[i].SetTitleFont(42, "XYZ")             # Font e dimensione di numeri e etichette degli assi 
   h_inv_mass[i].SetTitleSize(0.06, "XYZ")
   h_inv_mass[i].SetLabelFont(42, "XYZ")
   h_inv_mass[i].SetLabelSize(0.045, "XYZ") 
   h_inv_mass[i].GetXaxis().SetTitleOffset(1.0)      # Distanza tra l'etichetta e i numeri  
   h_inv_mass[i].GetYaxis().SetTitleOffset(1.5)
   h_inv_mass[i].SetMinimum(0.01)                    # Elimina lo zero dell'asse y che si sovrappone al 90 
   h_inv_mass[i].GetXaxis().SetTitle('m_{#gamma#gamma} [GeV]')    # Titoli degli assi
   h_inv_mass[i].GetYaxis().SetTitle('Events / GeV')
   onlyexp[i] = ROOT.TF1(f'onlyexp_{i}',"[0]*exp(-[1]*x)",100.,150.)          # Disegna il solo fondo in blu
   onlyexp[i].SetParameters(expgaus_per_mode[i].GetParameter(0), expgaus_per_mode[i].GetParameter(1))
   onlyexp[i].SetLineColor(ROOT.kBlue) 
   h_inv_mass[i].Draw("E1")                          # Disegna con errori
   onlyexp[i].Draw("SAME")
   h_inv_mass[i].Draw("E1 SAME")                      # ridisegna così la funzione blu è in secondo piano
```
